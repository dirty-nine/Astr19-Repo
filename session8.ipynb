```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 1, 100)
sinx = np.sin(x)
cosx = np.cos(x)
```

```python
fig, axs = plt.subplots(1, 2)
axs[0].plot(x, sinx)
axs[0].set_title("sin(x)")

axs[1].plot(x, cosx)
axs[1].set_title("cos(x)")

plt.savefig("session8Graph.png")
plt.show()
```

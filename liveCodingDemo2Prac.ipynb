```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 2*np.pi, 100)
y = 5 * np.sin(x)
```

```python
plt.plot(x, y)
plt.ylim(-1.1, 1.1)
plt.xlabel("x")
plt.ylabel("5sin(x)")
plt.gca().set_facecolor("gray")
plt.savefig("demo.png")
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0, 2*np.pi, 100)
y = 5 * np.sin(x)
plt.plot(x, y)
plt.ylim(-1.1, 1.1)
plt.xlabel("x")
plt.ylabel("5sin(x)")
plt.gca().set_facecolor("gray")
plt.savefig("demo.png")
```
